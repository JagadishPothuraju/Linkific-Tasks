```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)

data_path = Path("day5_dataset.csv")
if not data_path.exists():
    raise FileNotFoundError("Place your Day 5 dataset in this folder as day5_dataset.csv")

df = pd.read_csv(data_path)
print("Dataset shape:", df.shape)
display(df.head())
print("\\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))
display(df.describe(include="all").T)
```

## 1. Bar Chart — Total Sales by Category
A bar chart compares values across categories.

```python
sales_by_category = (
    df.groupby("Category", as_index=False)["Sales"]
      .sum()
      .sort_values("Sales", ascending=False)
)
sns.barplot(data=sales_by_category, x="Category", y="Sales")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
display(sales_by_category)
```

## 2. Line Chart — Sales Over Time
A line chart helps identify trends over time.

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
sales_over_time = df.groupby("Date", as_index=False)["Sales"].sum().sort_values("Date")
sns.lineplot(data=sales_over_time, x="Date", y="Sales", marker="o")
plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

## 3. Histogram — Distribution of Sales
A histogram shows the frequency of values in numeric ranges.

```python
sns.histplot(data=df, x="Sales", bins=8, kde=True)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## 4. Pie Chart — Share of Sales by Category
A pie chart displays each category's share of total sales.

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(7, 7))
plt.pie(category_sales, labels=category_sales.index, autopct="%1.1f%%", startangle=90)
plt.title("Share of Total Sales by Category")
plt.axis("equal")
plt.show()
```

## 5. Observation Summary

For the included example dataset, initial observations are:
