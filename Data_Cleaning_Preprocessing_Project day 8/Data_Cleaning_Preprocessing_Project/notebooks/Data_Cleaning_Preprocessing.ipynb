## Objective
Identify missing values, remove duplicates, standardize column names, convert incorrect data types, and save a cleaned dataset.

```python
from pathlib import Path
import pandas as pd
import numpy as np

data_path = Path("../data/raw_student_data.csv")
if not data_path.exists():
    data_path = Path("data/raw_student_data.csv")
df = pd.read_csv(data_path)
print("Original shape:", df.shape)
display(df.head())
```

## 1. Inspect missing values, data types, and duplicates

```python
print("Data types:")
display(df.dtypes.to_frame("Data Type"))
print("Missing values:")
display(df.isna().sum().to_frame("Missing Values"))
print("Exact duplicate rows:", df.duplicated().sum())
```

## 2. Standardize column names

```python
df.columns = (df.columns.str.strip().str.lower()
              .str.replace(r"[^a-z0-9]+", "_", regex=True)
              .str.strip("_"))
print(df.columns.tolist())
```

## 3. Convert incorrect data types

```python
for col in ["age", "marks", "attendance"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
display(df.dtypes.to_frame("Data Type"))
```

## 4. Remove duplicate rows

```python
before = len(df)
df = df.drop_duplicates().copy()
print("Duplicate rows removed:", before - len(df))
print("Current shape:", df.shape)
```

## 5. Handle missing values

```python
df["student_name"] = df["student_name"].fillna("Unknown")
df["email"] = df["email"].fillna("Not Provided")
for col in ["age", "marks", "attendance"]:
    df[col] = df[col].fillna(df[col].median())

df["department"] = df["department"].astype("string").str.strip().str.upper()
df["roll_number"] = df["roll_number"].astype("string").str.strip()
df["student_name"] = df["student_name"].astype("string").str.strip()
df["email"] = df["email"].astype("string").str.strip().str.lower()

print("Missing values after cleaning:")
display(df.isna().sum().to_frame("Missing Values"))
display(df.head())
```

## 6. Validate and save the cleaned dataset

```python
assert df["marks"].between(0, 100).all()
assert df["attendance"].between(0, 100).all()

output_path = Path("../outputs/cleaned_student_data.csv")
if not output_path.parent.exists():
    output_path = Path("outputs/cleaned_student_data.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)
print("Saved cleaned dataset to:", output_path)
print("Final shape:", df.shape)
display(df)
```

## Summary and Conclusion

- Identified missing values and exact duplicate rows.
- Standardized column names to lowercase snake_case.
- Converted age, marks, and attendance to numeric values.
- Converted invalid text values to missing values before imputation.
- Removed exact duplicate records.
- Filled missing text values with descriptive placeholders.
- Filled missing numeric values using the median.
- Validated marks and attendance ranges.
- Saved the cleaned dataset as `outputs/cleaned_student_data.csv`.

Data preprocessing improves data quality and makes datasets more suitable for analysis and machine learning.
