Practice NumPy arrays, indexing, slicing, mathematical operations, and list vs array comparison.

```python
import numpy as np
print('NumPy version:', np.__version__)
```

## 1D Array

```python
arr_1d = np.array([10, 20, 30, 40, 50])
print(arr_1d)
print('Dimensions:', arr_1d.ndim)
print('Shape:', arr_1d.shape)
```

## 2D Array

```python
arr_2d = np.array([[10,20,30],[40,50,60]])
print(arr_2d)
print('Dimensions:', arr_2d.ndim)
print('Shape:', arr_2d.shape)
```

## Indexing and Slicing

```python
print('First element:', arr_1d[0])
print('Third element:', arr_1d[2])
print('1D slice:', arr_1d[1:4])
print('2D element:', arr_2d[0,1])
print('2D slice:\n', arr_2d[:,1:3])
```

## Mathematical Operations

```python
marks = np.array([85,78,92,88,76])
print('Marks:', marks)
print('Sum:', np.sum(marks))
print('Mean:', np.mean(marks))
print('Maximum:', np.max(marks))
print('Minimum:', np.min(marks))
```

## Python List vs NumPy Array

```python
marks_list = [80,90,70,85]
marks_array = np.array(marks_list)
print('Python List:', marks_list)
print('NumPy Array:', marks_array)
print('Array + 5:', marks_array + 5)
```

## Conclusion
NumPy provides convenient array-based numerical operations that are widely useful in data analysis and AI/ML.
