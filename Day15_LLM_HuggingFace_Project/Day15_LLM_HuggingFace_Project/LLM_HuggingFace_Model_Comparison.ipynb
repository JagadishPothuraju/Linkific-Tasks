````python
import os, getpass, requests, pandas as pd
from IPython.display import display
HF_TOKEN = os.getenv("HF_TOKEN", "").strip()
if not HF_TOKEN:
    try: HF_TOKEN = getpass.getpass("Hugging Face token (hidden; Enter to skip): ").strip()
    except Exception: HF_TOKEN = ""
print("Token supplied." if HF_TOKEN else "No token supplied; API tests will be skipped until configured.")
````

## 4. Test three models
Example model IDs: GPT-2 for text generation, DistilBERT for sentiment analysis, and BART for summarization. Check model cards for current hosted inference support.

````python
API_URL = "https://api-inference.huggingface.co/models/{}"
HEADERS = {"Authorization": f"Bearer {HF_TOKEN}"} if HF_TOKEN else {}
tests = [
 {"task":"Text Generation","model":"gpt2","input":"Artificial intelligence can help students learn because","parameters":{"max_new_tokens":60,"return_full_text":False}},
 {"task":"Sentiment Analysis","model":"distilbert/distilbert-base-uncased-finetuned-sst-2-english","input":"The course was clear, practical, and very helpful.","parameters":{}},
 {"task":"Summarization","model":"facebook/bart-large-cnn","input":"Machine learning allows systems to learn patterns from data. Natural language processing helps computers work with human language. Large language models can generate, summarize, and classify text, but their outputs should be checked for errors.","parameters":{"max_length":60,"min_length":15,"do_sample":False}}
]
def query_hf(model, payload):
    if not HF_TOKEN: return {"status":"skipped","output":"No token supplied. Use the model page/provider or set HF_TOKEN."}
    try:
        r=requests.post(API_URL.format(model),headers=HEADERS,json=payload,timeout=90)
        try: data=r.json()
        except Exception: data=r.text[:500]
        return {"status":"success" if r.ok else f"HTTP {r.status_code}","output":data}
    except Exception as e: return {"status":"error","output":str(e)}
results=[]
for t in tests:
    payload={"inputs":t["input"]}
    if t["parameters"]: payload["parameters"]=t["parameters"]
    res=query_hf(t["model"],payload)
    results.append({"Task":t["task"],"Model":t["model"],"Input":t["input"],"Status":res["status"],"Output / Error":res["output"]})
results_df=pd.DataFrame(results)
display(results_df)
````
