## First and Last Five Rows

```python
df = pd.read_csv('../Dataset-Analysis/student_dataset.csv')
print(df.head())
print(df.tail())
```

## Rows and Columns

```python
print('Shape:', df.shape)
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
```

## Missing Values

```python
print(df.isnull().sum())
```

## Filtering

```python
print('Marks > 80:')
print(df[df['Marks'] > 80])
print('\nHyderabad students:')
print(df[df['City'] == 'Hyderabad'])
print('\nMarks > 80 and Age > 20:')
print(df[(df['Marks'] > 80) & (df['Age'] > 20)])
```

## Summary Statistics

```python
print(df.describe())
print('Average:', df['Marks'].mean())
print('Highest:', df['Marks'].max())
print('Lowest:', df['Marks'].min())
```
